## 8- Gere uma matriz 3x3 com números inteiros aleatórios e eleve todos os seus elementos ao quadrado.

```python
import numpy as np
import numpy.typing as npt

def matrixPow(pow: int, matrix: npt.NDArray[np.int_]) -> npt.NDArray[np.int_]:
    x: int = len(matrix[0])
    y: int = len(matrix)
    matrix = np.reshape(matrix, (1, x * y))[0]
    for index, _ in np.ndenumerate(matrix):
        matrix[index] **= pow
    return np.reshape(matrix, (y, x))

matrix: npt.NDArray[np.int64] = np.reshape(np.random.randint(1, 10, 9), (3, 3))
print(matrix)

matrixPow(2, matrix)
print(matrix)
```

## 9- Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.

```python
import numpy as np
import numpy.typing as npt

def normalizeBetween(min: float, max: float, array: npt.NDArray[np.float64]) -> npt.NDArray[np.float64]:
    new_array: npt.NDArray[np.float64] = np.array([])
    for value in array.astype(float):
        if value > min and value < max: new_array = np.append(new_array, value)
    return new_array

array: npt.NDArray[np.float64] = np.random.randn(10)
print("\nArray de números aleatórios:")
print(array)

array = normalizeBetween(0, 1, array)
print("\nArray normalizado entre 0 e 1:")
print(array)
print("")
```

## 10- Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.

```python
import numpy as np
import numpy.typing as npt

def countDiagonal(array: npt.NDArray[np.int_]) -> npt.NDArray[np.int_]:
    for index in range(len(array)):
        array[index][index] = index + 1
    return array

matrix: npt.NDArray[np.int_] = np.eye(6, dtype="i")

print(countDiagonal(matrix))
```

## 17- Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada valor por sua posição na matriz.

```python
import numpy as np
import numpy.typing as npt

def scaleByIndex(matrix: npt.NDArray[np.int_]) -> npt.NDArray[np.int_]:
    x: int = len(matrix[0])
    y: int = len(matrix)
    matrix = np.reshape(matrix, (1, x * y))[0]
    for index in range(len(matrix)):
        matrix[index] *= index + 1
    matrix = np.reshape(matrix, (y, x))
    return matrix

matrix: npt.NDArray[np.int_] = np.reshape(np.random.randint(1, 100, 16), (4, 4))
print(matrix)

matrix = scaleByIndex(matrix)
print(matrix)
```

## 18- Crie um array de 10 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 5 por 0.

```python
import numpy as np
import numpy.typing as npt

def setZeroDivisibleBy(divider: int, array: npt.NDArray[np.int_]) -> npt.NDArray[np.int_]:
    for index, value in np.ndenumerate(array):
        array[index] = 0 if value % divider == 0 else value
    return array

array: npt.NDArray[np.int_] = np.random.randint(1, 100, 10)
print(array)

array = setZeroDivisibleBy(5, array)
print(array)
```

## 20- Crie uma matriz de números inteiros e encontre o valor mínimo de cada linha.

```python
import numpy as np
import numpy.typing as npt

def generateNumber(times: int, start: int = 1, stop: int = 10) -> npt.NDArray[np.int_]:
    return np.random.randint(start, stop, times)

def generateLines(times: int, lineSize: int = 3) -> npt.NDArray[np.int_]:
    matrix: np.ndarray = np.array([])
    for _ in range(times):
        matrix = np.concatenate((matrix, generateNumber(lineSize, 1, 100)))
    return np.reshape(matrix, (times, lineSize))

def getLinesMin(array: npt.NDArray[np.int_]) -> npt.NDArray[np.int_]:
    linesLowest: npt.NDArray[np.int_] = np.array([])
    for line in array:
        linesLowest = np.append(linesLowest, min(line))
    return linesLowest

matrix: npt.NDArray[np.int64] = np.array(generateLines(7, 5))
print(matrix)

matrix = getLinesMin(matrix)
print("\nOs menores valores de cada linha são:")
print(matrix)
```

## 1- Calcule a integral da função f(x) = x²sin(x) no intervalo de 0 a π

```python
import scipy.integrate as integrate
import numpy as np

def integral(x: float):
    return (x ** 2) * np.sin(x)

resultado: float = integrate.quad(integral, 0, np.pi)
print("Resultado da integral: ", resultado)
```

## 2- Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
import scipy.integrate as integrate
import numpy as np

def differential_eq(x: int, y: int):
    return x - y

result = integrate.solve_ivp(differential_eq, [0, 11], [1])
print("Equação diferencial no intervalo entre 0 e 10:\n", result.y[0])
```

```python
import scipy.optimize as optimize

def func(x: int):
    return x**4 - 3*(x**3) + 2

result = optimize.root(func, [0.25])

print(result.x)
```

## 4- Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x+ 1.

```python
import scipy.optimize as optimize

def func(x: int):
    return -x + 4*x + 1

def oppositeFunc(x: int):
    return -func(x)

result = optimize.minimize(oppositeFunc, 0)

print(result.x)
```

## 14- Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.

## 17- Adicione uma coluna ao DataFrame que calcule o imposto (5%) sobre o valor da coluna Total.

```python
from pandas import DataFrame
import numpy as np

data = {
    'Produto': ['Produto 1', 'Produto 2', 'Produto 3', 'Produto 4', 'Produto 5'],
    'Preço': [2.99, 9.90, 5.50, 1.00, 0.50],
    'Quantidade': [14, 34, 50, 26, 39]
}

dataframe = DataFrame(data)

discount = np.vectorize(lambda price, stock : price * stock)
dataframe['Total'] = discount(data['Preço'], data['Quantidade'])

print(dataframe, "\n")

discount = np.vectorize(lambda total : total * 0.05)
dataframe['Imposto'] = discount(dataframe['Total'])

print(dataframe)
```

## 20- Agrupe os dados por uma coluna categórica e calcule a mediana de uma coluna numérica.

```python
from pandas import DataFrame
import numpy as np

data = {
    'Produto': ['Produto 1', 'Produto 2', 'Produto 3', 'Produto 4', 'Produto 5', 'Produto 6'],
    'Preço': [2.99, 9.90, 5.50, 1.00, 0.50, 0.05],
    'Quantidade': [np.random.randint(0, 100), np.random.randint(0, 100), np.random.randint(0, 100), np.random.randint(0, 100), np.random.randint(0, 100), np.random.randint(0, 100)],
    'Categoria': ['A', 'A', 'C', 'B', 'C', 'A']
}

dataframe = DataFrame(data)
print(dataframe, "\n")

categories = dataframe.groupby(by='Categoria')

tableData = {
    'Categoria': [],
    'Mediana Quantidade': []
}

for key, indexes in categories.indices.items():
    tableData['Categoria'].append(key)
    medianCategory: list[int] = []
    for index in indexes:
        medianCategory.append(dataframe['Quantidade'][index])
    tableData['Mediana Quantidade'].append(np.median(medianCategory))

table = DataFrame(tableData)

print(table)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 10, 100)
y = np.sin(x)

confidenceInterval = 0.95 * np.std(y)/np.sqrt(len(x))

plt.plot(x, y)
plt.fill_between(x, (y - confidenceInterval), (y + confidenceInterval), color='b', alpha=.15)
plt.show()
```
